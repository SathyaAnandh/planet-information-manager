```python
# FINAL PROJECT: Planet Information Manager


class Planet:
    def __init__(self, name, mass, gravity, distance):

        self.__name = name
        self.__mass = mass
        self.__gravity = gravity
        self.__distance = distance


    def get_name(self):
        return self.__name


    def display_info(self):
        print("\n--- Planet Information ---")
        print(f"Name       : {self.__name}")
        print(f"Mass       : {self.__mass} kg")
        print(f"Gravity    : {self.__gravity} m/s²")
        print(f"Distance   : {self.__distance} million km")
        print("---------------------------")



planet_list = []



def add_planet():
    print("\nEnter Planet Details:")
    name = input("Name: ")
    mass = input("Mass (kg): ")
    gravity = input("Gravity (m/s²): ")
    distance = input("Distance from Sun (million km): ")

    p = Planet(name, mass, gravity, distance)
    planet_list.append(p)
    print("\nPlanet added successfully! ")



def view_planets():
    if len(planet_list) == 0:
        print("\nNo planets found. Add some first!")
        return

    print("\n===== ALL PLANETS =====")
    for idx, planet in enumerate(planet_list):
        print(f"\nPlanet {idx + 1}:")
        planet.display_info()



def search_planet():
    search_name = input("\nEnter planet name to search: ").lower()

    for planet in planet_list:
        if planet.get_name().lower() == search_name:
            print("\nPlanet found!")
            planet.display_info()
            return

    print("\nPlanet not found.")



def delete_planet():
    del_name = input("\nEnter planet name to delete: ").lower()

    for planet in planet_list:
        if planet.get_name().lower() == del_name:
            planet_list.remove(planet)
            print("\nPlanet deleted successfully!")
            return

    print("\nPlanet not found. Try again.")



def menu():
    while True:
        print("\n==============================")
        print("     PLANET INFORMATION MANAGER")
        print("==============================")
        print("1. Add Planet")
        print("2. View All Planets")
        print("3. Search Planet")
        print("4. Delete Planet")
        print("5. Exit")
        print("==============================")

        choice = input("Enter your choice (1-5): ")

        if choice == "1":
            add_planet()
        elif choice == "2":
            view_planets()
        elif choice == "3":
            search_planet()
        elif choice == "4":
            delete_planet()
        elif choice == "5":
            print("\nExiting Planet Manager... Goodbye!")
            break
        else:
            print("\nInvalid choice! Please try again.")



menu()
```
